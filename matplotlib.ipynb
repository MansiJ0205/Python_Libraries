```python
import matplotlib.pyplot as plt
```

```python
%matplotlib inline
#run the matplotlib
```

```python
import numpy as np
```

```python
#create an array with start and stop values
x = np.arange(0,10)
y = np.arange(11,21)
```

```python
print(x)
print(y)
```

```python
plt.scatter(x,y,c='g')
plt.xlabel("x axis")
plt.ylabel("y axis")
plt.title("Graph in 2D")
```

```python
y=x*x
print(y)
```

```python
plt.plot(x,y)
```

```python
plt.plot(x,y,'r^', linestyle='dashed', linewidth=3,markersize=12)
plt.xlabel("x axis")
plt.ylabel("y axis")
plt.title("Graph in 2D")
```

```python
plt.subplot(2,2,1)
plt.plot(x,y,'r')
plt.subplot(2,2,2)
plt.plot(x,y,'g')
plt.subplot(2,2,3)
plt.plot(x,y,'b')
```

```python
x = np.arange(10,20)
y = 3*x + 5
plt.plot(x,y)
```

```python
#using sine functions in graph
print(np.pi)

x = np.arange(0, 4*np.pi, 0.1)
y = np.sin(x)

plt.title("sine wave curve")
plt.plot(x,y)
```

```python
# bar plot in matplotlib

x = [2,8,10]
y = [11,16,19]

x2 = [3, 9]
y2 = [5, 19]

plt.bar(x,y)
plt.bar(x2,y2, color='g')
```
